```python
# pickling
import pickle

# A live Python dictionary
user_data = {
    "username": "dev_mallem",
    "role": "backend_engineer",
    "skills": ["Python", "FastAPI", "Docker"],
    "is_active": True
}

with open('user_profile.pkl', 'wb') as file:
    pickle.dump(user_data, file)

print("Object successfully pickled and saved!")
```

```python
# 2. Unpickling (Deserialization)
import pickle

# Reading the byte stream back from the binary file
with open("user_profile.pkl", "rb") as file:
    loaded_data = pickle.load(file)

print(loaded_data)
print(type(loaded_data))  # Output: <class 'dict'>
print(loaded_data["skills"])  # Output: ['Python', 'FastAPI', 'Docker']
```

```python
# without *args
def Sum(a,b):
    return a+b

print(Sum(30,20))


# with *args
def ArgsSum(*args):
    # c=0
    # for i in args:
    #     c+=i
    # return c
    return sum(args)

print(ArgsSum(1,2,3,4,5,6))
```

```python
# **kwargs
def Robo(**kwargs):
    return kwargs
    

print(Robo(name='chitti', age=27, skill='ai'))
```

```python
def infinite_counters():
    num = 1
    while True:
        yield num
        num += 1

counter = infinite_counters()
print(next(counter))  # 1
print(next(counter))  # 2
print(next(counter))
for i in range(10):
    print(next(counter))
# The loop isn't running in the background; it only advances when we ask!
```
