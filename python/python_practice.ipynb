```python
# decorator
# a decorator is a function that modifies another function behavior without change its code

def log_call(func):
    def wrapper(*args, **kwargs):
        print(func.__name__)
        return func(*args, **kwargs)
    return wrapper

@log_call
def add(a,b):
    return a+b
print(add(10,20))
```

```python
# context manager
# It safely opens and closes resources like files or database connections.
class Door:
    def __enter__(self):
        print('open the door')
    def __exit__(self, a,b,c):
        print('exit the door')

with Door():
    print('Inside the room')
    print('im playing')
```

```python
# Identity operator
a=[1,2,3]
b=[1,2,3]
print(id(a))
print(id(b))
print(a is b)
print(a is not b)

z = a
print(a is z)
```
